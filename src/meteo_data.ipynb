```python
import cdsapi

c = cdsapi.Client()
c.retrieve(
    'reanalysis-era5-single-levels',
    {
        'product_type': 'reanalysis',
        'variable': ['mean_sea_level_pressure', '10m_u_component_of_wind', '10m_v_component_of_wind'],
        'year': [str(y) for y in range(2004, 2005)],
        'month': [f'{m:02d}' for m in range(1, 13)],
        'day': [f'{d:02d}' for d in range(1, 32)],
        'time': [f'{h:02d}:00' for h in range(24)],
        'area': [55, -3, 51, 4],   # North, West, South, East
        'format': 'netcdf',
    },
    'era5_thames_2004_2024.nc'
)
```

```python
import xarray as xr

ds = xr.open_dataset('era5_thames_2004_2024.nc')

# Southend Pier approx coordinates
lat, lon = 51.53, 0.71
point = ds.sel(latitude=lat, longitude=lon, method='nearest')

df_era5 = point.to_dataframe().reset_index()[['time', 'msl', 'u10', 'v10']]
df_era5.to_csv('southend_era5.csv', index=False)
```
